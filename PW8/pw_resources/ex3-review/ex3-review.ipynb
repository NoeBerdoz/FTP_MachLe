### d) A logistic regression can classify between 2 classes. How can we build a multi-class (with K classes) system with logistic regression ?

As mentioned in the course slides, a logistic regression is a one layer neural network where thetas are the weights and the activation function is a sigmoid. It has D inputs and one output which is $P(C_1 | \bold{x})$. 

To build a multi-class system we just need to implement the "one vs rest" approach. To do so we need a logistic regression for each of our classes. The first one is Class 1 vs the rest, the second is Class 2 vs the rest and so on until we get Class K vs the rest.

For regressor k, relabel the data as y = 1 for class k and y = 0 for every other class. The prediction is then argmax of all of the regressors.

We take the max value of all of these logistic regressors and that gives us the most probable class.

### e) Re-explain the link between logistic regression and neural networks. To what well-know loss function can be related the function we target to maximize in logistic regression ?

A logistic regression is a neural network with a single neuron and no hidden layer. Its inputs are the features $x_1, \dots, x_D$ plus a constant input $1.0$ for the bias, and its weights are the thetas, with $\theta_0$ as the bias weight. The neuron computes the weighted sum $z = \mathbf{x}\theta^T$ and applies a sigmoid activation, so its output is $g(z) = h_\theta(\mathbf{x}) = P(C_1 | \mathbf{x})$.

If we put the K "one vs rest" regressors from d) side by side, we get a 1-layer dense neural network with K neurons, where neuron k computes $P(C_k | \mathbf{x})$. The weights form a matrix of size $D \times K$, where column k is the theta vector of regressor k. All K neurons are trained at once with (stochastic) gradient methods, using a one-hot encoding of the labels as targets: for a sample of class k, the target is 1 for neuron k and 0 for all the others. Since the sigmoid outputs are trained independently, they don't necessarily sum to 1, which is why a softmax is usually added at the end of the network. The softmax normalizes the K outputs so that they form a probability distribution.

The objective function we maximize in logistic regression is

$$J(\theta) = \frac{1}{N} \sum_{n=1}^{N} y_{n} \log h_{\theta}(\mathbf{x_{n}}) + (1-y_{n}) \log (1 - h_{\theta} (\mathbf{x_{n}}))$$

This is exactly the cross-entropy loss used to train neural networks, with the opposite sign.

Maximizing $J(\theta)$ with gradient ascent is therefore the same as minimizing the cross-entropy with gradient descent. Both give the same thetas, and this is how neural network frameworks train their classifiers.
